```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import matplotlib.pyplot as plt


def heateq_fdm(a, b, T, N, M, theta):
    h = (b - a) / (N + 1)
    k = T / M
    e = np.ones(N)

    I = sp.eye(N, format="csc")
    G = h**(-2) * sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N).tocsc()

    B = (I + k * theta * G).tocsc()
    C = I - k * (1 - theta) * G

    x = np.linspace(a + h, b - h, N)
    u0 = x * np.sin(np.pi * x)

    f = -(1 - np.pi**2) * x * np.sin(np.pi * x) - 2 * np.pi * np.cos(np.pi * x)

    u = np.zeros((N, M + 1))
    err = np.zeros(M)

    u[:, 0] = u0

    for j in range(M):
        F = f * (
            theta * np.exp(-(j + 1) * k)
            + (1 - theta) * np.exp(-j * k)
        )

        u[:, j + 1] = spla.spsolve(B, C @ u[:, j] + k * F)

        exact = np.exp(-k * (j + 1)) * u0
        err[j] = np.linalg.norm(u[:, j + 1] - exact)

    return x, u, np.sqrt(h) * np.max(err)
```

```python
# parameters
a, b = 0.0, 1.0
T = 0.5
N = 8
M = 10
theta = 0.5   # Crank–Nicolson

x, u, max_err = heateq_fdm(a, b, T, N, M, theta)

print("Max error:", max_err)

# plot final time solution
plt.figure()
plt.plot(x, u[:, -1], label="Numerical")
plt.plot(x, np.exp(-T) * x * np.sin(np.pi * x),
         "--", label="Exact")
plt.xlabel("x")
plt.ylabel("u(x,T)")
plt.legend()
plt.title("Heat equation: numerical vs exact")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# fixed parameters
a, b = 0.0, 1.0
T = 0.5
theta = 0.5          # Crank–Nicolson
M = 400              # keep time error small

N_values = 10 * 2**np.arange(10)
h_values = []
errors = []

for N in N_values:
    x, u, max_err = heateq_fdm(a, b, T, N, M, theta)
    h = (b - a) / (N + 1)
    h_values.append(h)
    errors.append(max_err)

h_values = np.array(h_values)
errors = np.array(errors)

# reference line for O(h^2)
ref = errors[0] * (h_values / h_values[0])**2

plt.figure()
plt.loglog(h_values, errors, 'o-', label='Numerical error')
plt.loglog(h_values, ref, '--', label='O(h^2)')
plt.xlabel('h')
plt.ylabel('max error')
plt.legend()
plt.title('Spatial convergence')
plt.grid(True, which="both")
plt.show()

# optional: print observed slope
coeffs = np.polyfit(np.log(h_values), np.log(errors), 1)
print("Observed convergence rate:", coeffs[0])
```
